```python
import csv
```

```python
dat_file = "mprofile_20260809161435.dat"
csv_file = "memory_profile.csv"
```

```python
with open(dat_file, "r") as f_in, open(csv_file, "w", newline="") as f_out:
    writer = csv.writer(f_out)
    writer.writerow(["Time (s)", "Memory (MiB)"])   
    for line in f_in:
        if line.startswith("MEM"):                  
            parts = line.split()
            memory = float(parts[1])
            time   = float(parts[2])
            writer.writerow([time, memory])

print(f"Saved to {csv_file}")
```

```python
from collections import deque
```

```python
def read_last_n_lines(filepath, n=10):
    """
    Read the last n lines of a large file efficiently
    using a circular buffer (deque) — does NOT load
    the entire file into memory.

    Parameters:
        filepath : str  — path to the text file
        n        : int  — number of lines from the end
    """
    with open(filepath, "r", encoding="utf-8") as f:
        last_n = deque(f, maxlen=n)

    for line in last_n:
        print(line, end="")

# ── Usage ───────────────────────────────────────────────────
read_last_n_lines("C:\\Users\\xuean\\OneDrive - Imperial College London\\Documents\\PhDs\\SolvPred_notes\\SolvPred_JCIM_resubmission\\n_3_cand_35_01_01_01_0001\\log\\log_adv_filt_success_08092026161435.txt", n=80)
```
